```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
df = pd.read_csv("StudentsPerformance.csv")
```

 ------------------------------------------------------------
 1. Gender distribution
 ------------------------------------------------------------

```python
print("=== Gender Distribution ===")
print(df['gender'].value_counts())

print("\nPercentage:")
print(df['gender'].value_counts(normalize=True) * 100)
```


 ------------------------------------------------------------
 2. How many students passed the Math exam?
 ------------------------------------------------------------

```python
print("\n=== Math Exam Pass/Fail ===")
passed_math = df[df['math score'] >= 40].shape[0]
failed_math = df[df['math score'] < 40].shape[0]

print("Passed Math:", passed_math)
print("Failed Math:", failed_math)

def classify(score):
    if score >= 80: return 'A'
    elif score >= 70: return 'B'
    elif score >= 60: return 'C'
    elif score >= 50: return 'D'
    elif score >= 40: return 'E'
    else: return 'F'

df['math_grade'] = df['math score'].apply(classify)

print("\n=== Grade Distribution (A-F) ===")
grade_counts_m = df['math_grade'].value_counts().sort_index()
print(grade_counts_m)

print("\n=== Grade Percentages ===")
grade_percent_m = (df['math_grade'].value_counts(normalize=True) * 100).sort_index()
print(grade_percent_m)
```


  Math Grade Distribution (A–F)


```python
plt.figure(figsize=(8,5))
plt.bar(grade_counts_m.index, grade_counts_m.values, color='green')
plt.title("Math Grade Distribution (A–F)")
plt.xlabel("Grades")
plt.ylabel("Number of Students")
plt.grid(axis='y', linestyle='--', alpha=0.4)
plt.show()
```

 ------------------------------------------------------------
 3. Parental education level for pass/fail Math
 ------------------------------------------------------------

```python
print("\n=== Parental Education for Math Pass/Fail ===")
print("\nPass:")
print(df[df['math score'] >= 40]['parental level of education'].value_counts())

print("\nFail:")
print(df[df['math score'] < 40]['parental level of education'].value_counts())
```

 ------------------------------------------------------------
 4. How many students passed the Reading exam?
 ------------------------------------------------------------

```python
print("\n=== Reading Exam Pass/Fail ===")
passed_reading = df[df['reading score'] >= 40].shape[0]
failed_reading = df[df['reading score'] < 40].shape[0]

print("Passed Reading:", passed_reading)
print("Failed Reading:", failed_reading)

df['reading_grade'] = df['reading score'].apply(classify)

print("\n=== Grade Distribution (A-F) ===")
grade_counts_r = df['reading_grade'].value_counts().sort_index()
print(grade_counts_r)

print("\n=== Grade Percentages ===")
grade_percent_r = (df['reading_grade'].value_counts(normalize=True) * 100).sort_index()
print(grade_percent_r)
```


 Reading Grade Distribution (A–F)


```python
plt.figure(figsize=(8,5))
plt.bar(grade_counts_r.index, grade_counts_r.values, color='purple')
plt.title("Reading Grade Distribution (A–F)")
plt.xlabel("Grades")
plt.ylabel("Number of Students")
plt.grid(axis='y', linestyle='--', alpha=0.4)
plt.show()
```

 ------------------------------------------------------------
 5. Parental education for Reading pass/fail
 ------------------------------------------------------------

```python
print("\n=== Parental Education for Reading Pass/Fail ===")
print("\nPass:")
print(df[df['reading score'] >= 40]['parental level of education'].value_counts())

print("\nFail:")
print(df[df['reading score'] < 40]['parental level of education'].value_counts())
```

 ------------------------------------------------------------
 6. Students who passed/failed ALL subjects
 ------------------------------------------------------------

```python
print("\n=== Students Who Passed ALL Subjects (>=40) ===")
passed_all = df[(df['math score'] >= 40) &
                (df['reading score'] >= 40) &
                (df['writing score'] >= 40)]

print("Total passed all subjects:", passed_all.shape[0])

print("\nParental level of education for passed-all:")
print(passed_all['parental level of education'].value_counts())

print("\n=== Students Who Failed ALL Subjects (<40) ===")
failed_all = df[(df['math score'] < 40) &
                (df['reading score'] < 40) &
                (df['writing score'] < 40)]

print("Total failed all subjects:", failed_all.shape[0])

print("\nParental level of education for failed-all:")
print(failed_all['parental level of education'].value_counts())
```

 ------------------------------------------------------------
 7. The average score of students across all subjects
 ------------------------------------------------------------

```python
df["average_score"] = df[["math score", "reading score", "writing score"]].mean(axis=1)
df['average_grade'] = df['average_score'].apply(classify)
print(df[["math score", "reading score", "writing score", "average_score","average_grade"]])
```

 -------------------------------------------------------------------
 8. Parental education level for each grade based on all subjects
 -------------------------------------------------------------------

```python
print("\n=== Parental Education for Each Grade ===")
grades = ['A','B','C','D','E','F']

for grade in grades:
    print(f"\nGrade {grade} :")
    print(df[df['average_grade'] == grade]['parental level of education'].value_counts())


```


 Number of Students by Parental Education Level and Grade


```python
count_data = df.groupby(['parental level of education', 'average_grade']).size().reset_index(name='count')

plt.figure(figsize=(12, 6))
sns.set_style("whitegrid")
sns.barplot(
    data=count_data,
    x="parental level of education",
    y="count",
    hue="average_grade"
)

plt.xlabel("Parental Education Level")
plt.ylabel("Number of Students")
plt.title("Number of Students by Parental Education Level and Grade")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

 ------------------------------------------------------------
 9. Correlation between subjects
 ------------------------------------------------------------

```python
print("\n=== Correlation Matrix ===")
corr = df[['math score', 'reading score', 'writing score']].corr()
print(corr)
```

 ------------------------------------------------------------
 10. Test preparation effect
 ------------------------------------------------------------

```python
print("\n=== Test Preparation Course Impact ===")

prep = df.groupby('test preparation course')[['math score', 'reading score', 'writing score']].mean()
print(prep)
print("\nConclusion:")
print("Students who completed the test preparation tend to have a higher average score, "
      "while those who did not complete it scored slightly lower on average.")
```

 Making boxplots to relate the tests performance by test preparation

```python
fig, axes = plt.subplots(nrows=1, ncols=3, figsize=(18, 8))

sns.boxplot(x='test preparation course', y='math score', data=df, hue='test preparation course', ax=axes[0])
axes[0].set_title('Math Scores by Test Preparation Course')

sns.boxplot(x='test preparation course', y='reading score', data=df, hue='test preparation course', ax=axes[1])
axes[1].set_title('Reading Scores by Test Preparation Course')

sns.boxplot(x='test preparation course', y='writing score', data=df, hue='test preparation course', ax=axes[2])
axes[2].set_title('Writing Scores by Test Preparation Course')

plt.show()
```
